# as we can see there is some outlier so lets remove it

there are several possible ways to remove it like iqr method(INTER QUANTILE RANGE) ,z-score method.but we use iqr method.

```python
q1=data["trestbps"].quantile(0.25)
q1
```

```python
q3=data["trestbps"].quantile(0.75)
q3
```

```python
IQR=q3-q1
IQR
```

```python
min_range=q1-(1.5*IQR)
max_range=q3+(1.5*IQR)
min_range ,max_range
```

```python
data=data[data["trestbps"]<=max_range]
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="trestbps",data=data)
plt.show()
```

```python
sns.distplot(data["trestbps"])
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=data.iloc[:,:-1]
```

```python
y=data["target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

# now analyze the data using confusion matrix to understand how much accurate is the model

```python
from sklearn.metrics import confusion_matrix,precision_score,recall_score,f1_score
```

```python
cf=confusion_matrix(y_test,lr.predict(x_test))
cf
```

```python
sns.heatmap(cf,annot=True)
plt.show()
```

```python
recall_score(y_test,lr.predict(x_test))*100
```

```python
precision_score(y_test,lr.predict(x_test))*100
```

```python
f1_score(y_test,lr.predict(x_test))*100
```
